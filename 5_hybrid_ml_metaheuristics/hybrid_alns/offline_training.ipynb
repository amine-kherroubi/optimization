# Offline Repair Model Training Notebook

This notebook is dedicated to training the offline machine-learning repair model used by the hybrid ALNS solver.


## 1) Goal

Train `repair_model.pkl` from ALNS state/label data generated by `collect_alns_states.py`.

Why offline training:
- Keeps the optimization loop lightweight at runtime.
- Enables experimentation with model hyperparameters independently from search.


## 2) Recommended command-line training setup

These arguments prioritize strong generalization and stable ranking quality for repair decisions:
- `--model xgb`: robust gradient-boosted trees for tabular features.
- `--n-estimators 400`, `--max-depth 6`, `--learning-rate 0.05`: good bias/variance balance.
- `--subsample 0.9`, `--colsample-bytree 0.8`: regularization for better out-of-sample performance.
- `--seed 42`: reproducibility.


```python
!python collect_alns_states.py \
    --instances-dir ../instances \
    --output states_dataset.csv \
    --n-instances 500 \
    --iters 300 \
    --seed 42
```

```python
!python train_repair_model.py \
    --data states_dataset.csv \
    --out repair_model.pkl \
    --model xgb \
    --n-estimators 400 \
    --max-depth 6 \
    --learning-rate 0.05 \
    --subsample 0.9 \
    --colsample-bytree 0.8 \
    --seed 42
```
